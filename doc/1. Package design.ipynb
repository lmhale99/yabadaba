# Introduction to yabadaba: Package design

The design of the yabadaba package focuses on modularity. To this end, the major components of yabadaba are represented by classes and objects and can be easily extended by defining new subclasses. This Notebook outlines the base class components, which are described in more detail in the subsequent Notebooks.

- __Database__ specifies common APIs for interacting with databases of different infrastructures. Each Database subclass implements the common APIs for a specific database infrastructure. Objects of a Database subclass then are initialized with database access parameters allowing for it to interact with a database of that infrastructure.
- __Record__ specifies how to interact with data associated with a particular schema or schema subset. Each Record subclass is defined specific to a schema. Objects of a Record subclass then interpret a single data entry of that schema.
- __Value__ specifies how individual data values are to be interpreted. Each Value subclass is associated with a different data type, which can be simple types like strings or integers or representations of more complex data. Value objects are used to define the schemas associated with a Record subclass.
- __Query__ specifies database query operations that can be performed related to a Value in a Record. Each Query subclass defines a database query operation and contains methods for efficiently implementing that action in the different database architectures. Query objects are defined associated with Value objects. Most basic Value subclasses automatically create one default Query associated with it.
- __ModuleManager__ manages the subclasses of the above class categories in a modular way allowing for extension packages to dynamically add more subclasses as they are imported. When yabadaba is initialized it automatically creates four ModuleManagers, one for each of the above class categories.
- __UnitConverter__ provides a means of handling unit conversions with minimal overhead. One Unitconverter object yabadaba.unitconvert is created when yabadaba is initialized and should almost always be used rather than creating a new object.
- __Settings__ provides an interface for saving and loading custom settings and options for later Python sessions. One Settings object yabadaba.settings is created when yabadaba is initialized and typically should be used rather than creating a new object.
