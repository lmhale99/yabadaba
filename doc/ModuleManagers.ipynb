# Introduction to yabadaba: ModuleManagers

Yabadaba has been designed to be modular and easily extendable, both within yabadaba itself and by importing other packages that extend yabadaba.  This modularity is largely achieved through the use of ModuleManager objects. 

## 1. ModuleManager usage

The typical way of using ModuleManagers is to import the existing manager objects from yabadaba and interacting with them rather than creating new ModuleManager objects. This allows for the managers to retain any modular components defined by yabadaba itself or any extension packages. Yabadaba contains four ModuleManager objects (databasemanager, recordmanager, valuemanager, and querymanager), each of which supports the modular handling of the associated Class categories.

### 1.1. Check the status of modules

You can check the status of the module styles using these convenience methods

- __check_styles()__ prints information about the current module styles that successfully and unsuccessfully loaded.
- __assert_style()__ takes a style and raises an error if it was not loaded or failed to load.
- __loaded_style_names__ is the list of module styles that successfully loaded.
- __failed_style_names__ is the list of module styles that unsuccessfully loaded.
- __loaded_styles__ is a dict containing the successfully loaded modules by their styles.
- __failed_styles__ is a dict containing the error messages raised by the unsuccessfully loaded modules by their styles.

### 1.2. Fetch/build a module by style

- __get_class()__ retrieves the loaded module class associated with a style.
- __init()__ initializes a new object of a loaded module class. It takes the module style as well as any initialization args and kwargs.

### 1.3. Add new modules

- Directly add modules to the __loaded_styles__ if they were defined locally.
- __import_style()__ tries to import a new module style. If successful, the module is added to loaded_styles. If unsuccessful, the error message is added to failed_styles.  It takes the following parameters
    - style (*str*) The style name to associate with the modular class.
    - modulename (*str*) The name of the module to try to import.
    - package (*str, optional*) The name of the package which is to act as the anchor for resolving relative package names.
    - classname (*str, optional*) The name of the class in the imported module to associate with the style.  If not given, will use the final name of the modulename path.

The modulename and package parameters correspond to the similarly named parameters for importlib.import_module(). An example of using this found in the yadabada_demo/record/\_\_init\_\_.py file within the doc directory:

```Python
# style: 'album'
# modulename: '.album' module is relative to package
# package: __name__ == 'yadabada_demo.record'
# classname: 'Album' class is within the module
# Thus, style 'album' points to the class yadabada_demo.record.album.Album
recordmanager.import_style('album', '.album', __name__, 'Album')
```

## 2. New ModuleManagers

While most common usage will simply use the pre-defined ModuleManager objects, you can create more if you wish. Two use cases that I've seen are
- Manage additional subclass categories in a similar way with their own ModuleManager objects.
- Replace one of the default managers so that the new manager points to "extended" versions of the subclasses rather than the default versions.

When you initialize a new ModuleManager, it only takes a single input "parentname" which is the name of the parent class that all modules are subclassed from. Currently, this value is only used for messages and metadata checks.
